```python
from langchain.chat_models import init_chat_model

model=init_chat_model(
    model="gemini-3.5-flash-lite",
    model_provider="openrouter"
)
model
```

```python
response=model.invoke("why do parots talk?")
response.content
```

### Streaming n Batching

Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time: - ".stream"

```python
for chunk in model.stream("why parots talk?? "):
    print(chunk.text, end="|" , flush =True)
```

### BATCH

a batch is a collection of independent requests to a model - that can significantly improve performance & reduce costs - as the processing can be done in parallel.  === ".batch"

```python
responses= model.batch([
    "Why do parrots have colorful feathers",
    "why airplanes fly?",
    "What is the difference between a crocodile and an alligator?"]

)

for response in responses:
    print(response.text)
```

```python
responses= model.batch([
    "Why do parrots have colorful feathers",
    "why airplanes fly?",
    "What is the difference between a crocodile and an alligator?"],

    config={
        "max_concurrency" : 5,
        "max_tokens" : 2000
    }
)

for response in responses:
    print(response.text)
```
